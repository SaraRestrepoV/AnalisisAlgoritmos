Ejercicio 5 – Fusionar Intervalos (LeetCode 56)

https://leetcode.com/problems/merge-intervals/

> Enunciado

Dado un arreglo de intervalos `intervalos[i] = [inicio_i, fin_i]`, fusionar todos
los intervalos que se solapan y retornar una lista sin solapamiento que cubra
exactamente los mismos puntos que la entrada.

Estrategia voraz Greedy: ordenar los intervalos por su punto de inicio y, en una sola
pasada, decidir localmente si el intervalo actual se fusiona con el último ya aceptado.

> Análisis de complejidad

> Tiempo: O(n log n)
Todo el costo viene del ordenamiento al inicio. Lo que viene después,
recorrer la lista una sola vez, es O(n) y no cambia el resultado.

> Espacio: O(n)
La lista `resultado` en el peor caso guarda los n intervalos originales
si ninguno se solapa con otro, así que crece proporcional a la entrada.

> n = número de intervalos en la entrada.


```python
from typing import List

class Solution:
    def merge(self, intervalos: List[List[int]]) -> List[List[int]]:
        # 1. Ordenar por inicio
        intervalos.sort(key=lambda x: x[0])
        
        resultado = [intervalos[0]]
        
        # 2. Recorrer el resto
        for inicio, fin in intervalos[1:]:
            ultimo_fin = resultado[-1][1]
            
            if inicio <= ultimo_fin:
                resultado[-1][1] = max(ultimo_fin, fin)
            else: 
                resultado.append([inicio, fin])
        
        return resultado


# ── Casos de prueba ───────────────────────────────────────────────────────────
sol = Solution()

# Ejemplo 1 
assert sol.merge([[1,3],[2,6],[8,10],[15,18]]) == [[1,6],[8,10],[15,18]]

# Ejemplo 2
assert sol.merge([[1,4],[4,5]]) == [[1,5]]

# Ejemplo 3
assert sol.merge([[4,7],[1,4]]) == [[1,7]]

# Caso borde: un solo intervalo
assert sol.merge([[5,5]]) == [[5,5]]

# Todos disjuntos
assert sol.merge([[1,2],[3,4],[5,6]]) == [[1,2],[3,4],[5,6]]

# Todos solapados
assert sol.merge([[1,10],[2,8],[3,6]]) == [[1,10]]

print("Todos los casos pasaron correctamente.")
```
