# 416. Partition Equal Subset Sum

## Descripción del problema
Dado un arreglo de números enteros, debemos determinar si es posible dividirlo en dos subconjuntos cuya suma sea igual.

## Programación dinámica

Definimos un arreglo:

`puede_formar[s]`: indica si es posible formar la suma `s` con algunos elementos del arreglo.

### Caso base
`puede_formar[0] = True` (siempre se puede formar suma 0)

### Transición
Para cada número:
`puede_formar[s] = puede_formar[s] or puede_formar[s - numero]`


```python
def puede_particionar(numeros):
    suma_total = sum(numeros)
    
    if suma_total % 2 != 0:
        return False
    
    suma_objetivo = suma_total // 2

    puede_formar = [False] * (suma_objetivo + 1)
    puede_formar[0] = True
    
    for numero in numeros:
        for suma_actual in range(suma_objetivo, numero - 1, -1):
            if puede_formar[suma_actual - numero]:
                puede_formar[suma_actual] = True
    
    return puede_formar[suma_objetivo]
```

```python
# Casos de prueba

caso_1 = [1, 5, 11, 5]
caso_2 = [1, 2, 3, 5]

print("Caso 1:", puede_particionar(caso_1))
print("Caso 2:", puede_particionar(caso_2))
```
